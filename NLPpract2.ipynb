```python
import nltk
from collections import Counter
from nltk.util import ngrams


text = "I love NLP and I love Python"
words = text.lower().split()


unigrams = list(ngrams(words, 1))
bigrams = list(ngrams(words, 2))
trigrams = list(ngrams(words, 3))


unigram_counts = Counter(unigrams)
bigram_counts = Counter(bigrams)
trigram_counts = Counter(trigrams)



print("--- Unigram Probabilities ---")
total_words = len(words)
for unigram, count in unigram_counts.items():
    probability = count / total_words
    print(f"{unigram[0]}: {probability:.4f}")

print("\n--- Bigram Probabilities ---")
for bigram, count in bigram_counts.items():

    first_word_tuple = (bigram[0],)
    denominator = unigram_counts[first_word_tuple]

    probability = count / denominator
    print(f"{bigram[0]} -> {bigram[1]}: {probability:.4f}")

print("\n--- Trigram Probabilities ---")
for trigram, count in trigram_counts.items():

    first_two_words_tuple = (trigram[0], trigram[1])
    denominator = bigram_counts[first_two_words_tuple]

    probability = count / denominator
    print(f"{trigram[0]} {trigram[1]} -> {trigram[2]}: {probability:.4f}")
```
